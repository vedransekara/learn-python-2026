# Recap - Functions
### How does Python know which argument corresponds to which parameter?

````python
# a function can have several parameters
def add_numbers(x, y, z):
    my_sum = x + y + z
    return my_sum
# parameters in function def: x, y, z

# in this function call
add_numbers(1, 2, 3)
# python knows BY POSITION: 
# first argument (1) is x 
# second argument (2) is y
# third argument (3) is z
````

````python
# a function can have several parameters
def add_numbers(x, y, z):
    my_sum = x + y + z
    return my_sum

# function call
# we can explicitly "assign the arguments by parameter name"
add_numbers(x = 1, y = 2, z = 3)
````

````python
# why is it useful to assign arguments by NAME?
# because ORDER MATTERS!
def compute_power(x, y):
    return x**y

print(compute_power(5,2)) # positional arguments >> computes 5 * 5
print(compute_power(2,5)) # positional argument >> computes 2 * 2 * 2 * 2 * 2
````

````python
# we can have "default values" for "optional parameters":
def compute_power(x, y = 2):
    return x ** y

# if we don't provide an argument for the second parameter (y), it takes on the default value
compute_power(3)

# if we DO provide an argument for the second parameter it y takes on that input
# compute_power(3, 4)
````

````python
# we can have "default values" for "optional parameters":
def compute_power(x, y = 2):
    # x is required (must be provided)
    # y is optional (if not provided, gets the default value)
    return x ** y

# compute_power(3) # if we DO NOT specify y, it will return x ** 2
# compute_power(3,3) # if we DO specify y, it will return x ** y
# compute_power(y=3) # if we DO NOT specify x, it will complain
````

````python
# RULE FOR FUNCTION DEFINITIONS:
# required parameters must ALWAYS COME FIRST
# and default parameters come LAST.
# this will throw an error message:
def compute_power(x = 10, y): # THIS IS WRONG
    return x ** y
````

# What is a list?
* A list is an (versatile) collection or containier of values (called 'elements' or 'items') in an __ordered sequence__
* i.e., there is a "first", "second", …, "last" element
* lists begin with an opening bracket `[` and end with a closing bracket `]` (or using the built-in function `list`)
* in between it can contain other values, which are separated by comma

# since a `list` is an collection of values, it can:
* be assigned to a variable
* passed as an argument to a function
* be returned by a function

# Getting a Part of a List using Slices
#### - Form: `animals[i:j]` where `i` and `j` are integers
#### - `i` says where the slice starts
#### - `j` says where the slice ends (up to, but not including `animals[j]`)
#### - `i` and `j` can be skipped, e.g., `animals[:]` is valid and selects the entire list
#### - <font color='orange'>(if `i` in `animals[i:j]` is left out the slide starts from index 0 by default, if `j` is left out slice ends at last element by default) </font>

````python
animals = ['cat', 'bat', 'rat', 'elephant']
animals[2]
````

````python
# slicing list to get elements from 0'th index to 2nd index (but not incuding the value)
animals[0:2]
````

````python
# equivalent to use 
animals[:2]
# if you don't specify start i it assumes you take from start of list
````

````python
# different variations
animals = ['cat', 'bat', 'rat', 'elephant']
print(animals[0:4])
print(animals[1:3])
print(animals[0:-1])
print(animals[1:])
````

# Other data types - tuples - `tuple`!
### lists are called mutable (meaning you can edit/change them)
### tuples are like lists, but they are immutable (meaning you cannot change them!) - <font color='red'> good for fixed data </font>

````python
# a list containing three elements
l = ['a','b','c']

# a tuple containing three elements
t = ('a','b','c') # notice the regular brackets
````

````python
# or 
# a list containing three elements
l = list(['a','b','c'])

# a tuple containing three elements
t = tuple(('a','b','c')) # notice the regular brackets
````

````python
# these things are equivalent
('a', 'b', 'c') == tuple(('a', 'b', 'c'))
````

````python
# you can index and slice tuples in the same way as lists
print(l[2])
print(t[2])
````

````python
# you can index and slice tuples in the same way as lists
print(l[:2])
print(t[:2])
````

````python
# you can add extent list and tuples in the same way
print([1,2,3,4,5]+[6,7,8])
print((1,2,3,4,5)+(6,7,8))
````

````python
# but you cannot edit tuples
l[0] = 'q'
t[0] = 'q'
````

````python
# or delete objects in tuples - you cannot change the sequence in any way
del l[0]
del t[0]
````

````python
# you can easily switch between lists and tuples types
print(tuple(['a','b','c']))

print(list(('a','b','c')))
````

### list are great for ordered data where you want the flexibility of changing things
### tuples are fantastic for ordered data where nothing should change

### Last but not least, `.format()` a very powerful tool for formatting strings

````python
# so far we have inserted things into string in a very silly way
x = 3
print('Today I have eaten ' + str(x) + ' ice creams, and you have eaten ' + str(4) )
````

````python
# .format() is a different way of inserting variable values into literal text

# by uorder:
print('Today I have eaten {} ice creams, and you have eaten {}'.format('3', '4')) # by order (with empty curly braces)

# by position:
print('Today I have eaten {1} ice creams, and you have eaten {0}'.format('4', '3'))

# by variable name:
print('Today I have eaten {x} ice creams, and you have eaten {y}'.format(y = '4', x = '3')) # by variable name
````
