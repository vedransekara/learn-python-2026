## Logical operators in Python

```python
not, and, or 
``` 

Python evaluates combinations of logical operators in the following order of precedence:

`not > and > or`

You can always use round brackets to explicitly state the order of evaluation: 
```python 
p and not q or s # same as: (p and (not q)) or (s)
```

Note that the **Python** `or` in Python is **inclusive**: `p or q` means either p, or q, **or both**. 

> This is different from the **exclusive** `or` (XOR) in logic: `p XOR q` means either p, or q, **but not both**.

# Conditional statements in Python

`if`, `else`, `elif`

## the form of an `if` statement

```python
# "condition" is evaluated to either True or False
if condition: # note the : 
    expression1 # note the indent
```

Read as "If the condition is true, then execute the statements in the clause; 
otherwise skip the clause"

* Note the colon `:` at the end!
* Note that the next line starts with an indented block `    print("Hello!")`

````python
be_friendly = True
if be_friendly == True: # keyword "if" + condition + :
    print("Hello!") # indented line of code after "if"
````

```python
be_friendly = False
if be_friendly == True: # condition: be_friendly. if that condition is False, then:
    print("hello")  # DO NOT execute the INDENTED lines of code below
```

````python
be_friendly = False
if be_friendly == True:
    print("Hello!")
print("The if-statement ended in the line above.")
````

````python
be_friendly = True
if be_friendly == True:
    print("Hello!") # the indented lines are executed only if the if-condition evaluates to True
print("The if-statement ended in the line above.") # the non-indented lines that follow are executed in any case
````

## `else` statement

__`else` can only follow an if statement__
```python
# "condition" is evaluated to either True or False
if condition:
    expression1 # note the indent
else:
    expression2 # note the indent
```

Works as "If the condition is true,
execute the statements in the `if`-clause, otherwise execute the
statement in the `else`-clause”

````python
be_friendly = True
if be_friendly == True:
    print("Hello!") # the indented lines are executed only if the if-condition evaluates to True
else:
    print(":-/")
print("The if-else-statement ended in the line above.") # the non-indented lines that follow are executed in any case
````

# `elif`: introducing one more condition

```python
if condition1:
    expression1
elif condition2:
    expression2
else:
    last_expression
```

With if/else statements, we have only two possible outcomes.
`elif` ("else if") allows us to declare more clauses that we can execute
* First condition that evaluates to `True` is taken
* if all conditions are False, the else clause is taken

````python
# Let's say we want to write a slightly more sophisticated "email reply generator":
# friendliness level 0-4: "ok"
# friendliness level 5-8: "Ok, thanks!"
# friendliness level 9+ : "Alright! Many thanks and have a wonderful day"
````

````python
# A slightly more sophisticated "email reply generator":
# friendliness level 0-4: "ok"
# friendliness level 5-8: "Ok, thanks!"
# friendliness level 9+ : "Alright! Many thanks and have a wonderful day"

friendliness_level = 5

if friendliness_level > 8:
    print("Alright! Many thanks and have a wonderful day")
elif friendliness_level > 4:
    print("Ok, thanks!")
else:
    print("ok")
    
print("done")
````

### Note that the first condition that evaluates to `True` is taken
Lets see what happens if we revere the order of our conditions

````python
friendliness_level = 8 # lets raise this a bit

if friendliness_level > 4: # this part was the elif statement before
    print("Ok, thanks!")
elif friendliness_level > 8: # this part was the if statement before
    print("Alright! Many thanks and have a wonderful day")
else:
    print("ok")
    
print("done")
````
