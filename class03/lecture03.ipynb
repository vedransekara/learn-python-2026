# While loops in Python

### makes a block of code execute over and over again, as long as a condition is true (looks similar to an `if` statement)

```python
while condition:
    clause
```

### looks the same as an `if` statement, but there is a different in behavior
```python
if condition:
    clause
```

## Lets recap if

````python
i = 5
if i > 3:
    print('hello 👋')
# if the condition is True the clause is run once
````

## what does while do?
here `i>5` is the condition and `print('hello 👋')` and `i = i + 1` are the clause

````python
i = 0
while i < 5:
    print('hello 👋')
    i = i + 1
````

### Be careful with while loops, you can create infinite loops!

````python
# you can make our computer work FOREVER
# the code after "while" is executed WHILE (AS LONG AS) the condition is True.
x = 5
while x == 5: # the condition x == 5 never changes
    print("i'm stuck in a while loop!")
````

## There is also another way of breaking out of while loops
### `break` and `continue` statements
Consider these two programs which produce identical behavior

````python
while True: # always true - basicall a forever loop
    print('Please input a fruit:')
    fruit = input()
    if fruit == 'banana':
        break # break out of the loop
    print('On no, wrong fruit!')
        
print('Thank you 🍌')
````

````python
fruit = ''
while fruit != 'banana':
    print('Please input a fruit:')
    fruit = input()
    
print('Thank you 🍌')
````

### `break` Immediately terminates a loop when a specified condition is met. After the loop exits, program execution continues with the next statement following the loop

***
## `continue` statement
### `continue` is a loop control statement that skips the rest of the code inside the loop for the current iteration and moves to the next iteration immediately.

````python
while True:
    print('Who are you?')
    name = input()
    if name != 'Joe':
        continue
        
    print('Hello Joe. What is the password?') # skipped by continue if name is not Joe
    password = input()                        # skipped by continue if name is not Joe
    if password == 'banana':                  # skipped by continue if name is not Joe
        break                                 # skipped by continue if name is not Joe
````

## There is also a statement called `pass`
### It does nothing when executed! 👀
It is often used as a placeholder - e.g. used to keep code blocks valid where a statement is required but no logic is added yet

````python
i = 0
while i < 10:
    if i >= 5:
        pass
    else:
        print('hello 👋')
    i += 1
````

## one last thing ... 🫠
### `break`, `continue`, and `pass` only affect the loop you are in. 
### If you have nested loops you might still be in trouble

````python
i = 0
while i < 3:
    j = 0
    while True     # break only affects this loop
        print(i,j) # break only affects this loop
        if j > 2:  # break only affects this loop
            break  # break only affects this loop
        j += 1     # break only affects this loop
    i += 1
````

## a note about the `in` operator in the `for` loop

### in the below example `in` is just a marker that separates the loop-index variable (`i`) from whatever you're looping over (`range(5)`).

```python
for i in range(5):
    some_code
```

### Generally, the `in` operator in Python (when not used in a `for` loop) is a membership operator which check if a value is part of a sequence or collection (don't worry about that these are, we'll define them in detail later).

````python
# example, here python checks if 'a' is part of the string 'hello'
'a' in 'hello'
````

````python
# example, here python checks if 'a' is part of the string 'banana'
'a' in 'banana'
````

### Remember its case sensitive 

````python
# example, here python checks if 'a' is part of the string 'BANANA'
'a' in 'BANANA'
````

### you can also use logical operators in combination with `in`

````python
# we can also use the logical operators to negate this
'a' not in 'banana'
````

## if you don’t want to count from 0

```python
for i in range(start, stop, step):
    clause
```
* `start` starting integer (optional, default is `0`)
* `stop` stopping integer (not included, stops before this integer)
* `step` an integer specifying the step size (optional, default is `1`)

````python
# start = 0, end = 10 (not included), step = 2
for i in range(0,10,2):
    print(i)
````

### increments can be negative! (and so can start, and stop values)

````python
start = -10
end = - 15
step = -1
for i in range(start, end, step):
    print(i)
````

# `For`-loops "leak"
### The current value of the loop variable will "leak" (become available) once the for-loop ends.

````python
for i in range(5):
    print('in for-loop:', i) 

# the last value of x has "LEAKED":
print('after for-loop:', i)
````

## The loop variable "leaks"
### ... even if it was defined with a different value before!

````python
# THE LOOP VARIABLE "LEAKS"
# ... even if it was defined with a different value before!
i = "efghij"
print('before for-loop:',i)

for i in range(5):
    print('in for-loop:', i) 

# the last value of loop_var has "LEAKED":
print('after for-loop:', i)
````

## you can have nested loops (both `for` and `while`)

````python
for i in range(3):
    for j in range(3):
        # print i, j, and their sum
        print(i,j,i+j)
````

# if there is time lets analyze a program for playing the hangman game 

````python
# my first python program
print('Time to play Hangman!')
print('Start guessing...')

# ---------------------------------------------------------------------
# super secret code to load the super secret name in a super secret way
with open('secret_name.txt', 'r') as f:
    secret_name = f.readlines()[0]
# ---------------------------------------------------------------------
guesses = ''
turns = 10

while turns > 0:
    failed = 0

    for char in secret_name:
        if char in guesses:
            print(char, end='')
        else:
            print('_',end='')
            failed += 1

    if failed == 0:
        print('\nYou won!')
        break

    print()
    guess = input('Guess a character: ')

    if len(guess) > 1:
        print('Only one charracter at the time, please!')

    guesses += guess

    if guess not in secret_name:
        turns -= 1
        print('Wrong\n')
        print('You have',turns,'more gueses')
        if turns == 0:
            print('You lose!')
````
