## Understanding tasks
### Q1 What is the difference between range(10), range(0, 10), and range(0, 10, 1) in a for loop?

They all do the same thing. The range(10) call ranges from 0 up to (but not including) 10,
range(0, 10) explicitly tells the loop to start at 0, and range(0, 10, 1) explicitly tells the
loop to increase the variable by 1 on each iteration.

### Q2: Write a short program that prints the numbers 1 to 10 using a for loop. Then write an equivalent program that prints the numbers 1 to 10 using a while loop

````python
# code for 'for loop' here
for i in range(1, 11):
    print(i)
````

````python
# code for 'while loop' here
i = 1
while i <= 10:
    print(i)
    i = i + 1
````

### Q9: Summing numbers
This question consists of 3 tasks (T).
* Task 1: Use a `for` loop and the variable `my_sum` (initially set to 0) to calculate the sum of all numbers from 0
to 500.

````python
# your code here
my_sum = 0
for i in range(0,500+1): # you can also use range(501)
    my_sum += i
````

* Task 2: Use a `for` loop and the variable `my_sum` (initially set to 0) to calculate the sum of all even numbers
from 0 to 500.

(hint: Use the modulo operator `%` to find even numbers)

````python
# your code here
my_sum = 0
for i in range(0,500+1): # you can also use range(501), or range(0,501,2)
    if i % 2 == 0: # finds all even numbers
        my_sum += i
````

* Task 3: Use a `for` loop and the variable `my_sum` (initially set to 0) to calculate the sum of all divisible by 7
numbers from 0 to 500.

(hint: Use the modulo operator `%`)

````python
# your code here
my_sum = 0
for i in range(0,500+1): # you can also use range(0,501,7)
    if i % 7 == 0:
        my_sum += i
````

````python
# your code here
customers = 0
while True:
    action = input("What has happened? ")
    if action == "enter" and customers == 20:
        print("No one can enter analog because it is full.")
        continue
    if action == "enter":
        customers += 1
    elif action == "leave":
        customers -= 1
    elif action == "close":
        break
    else:
        print("I did not understand that. Please repeat.")
        continue
        
    increment_counter = 0
    for i in range(20):
        if i < customers:
            print("O", end="")
        else:
            print("-", end="")
        increment_counter += 1
        
        if increment_counter == 5:
            print(" ", end="")
            increment_counter = 0
    print()
    
    if customers == 20:
        print("BARRICADE THE DOORS!")
````
