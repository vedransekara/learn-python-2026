# Part 1: Guest book
Write a program that uses a while loop that prompts users for their name (i.e. using `input()`). When they enter their name, print
a greeting to the screen and add a line recording their visit in a file called `guest_book.txt`. Make sure
each entry appears on a new line in the file by opening it in a text editor

__suggested solution__

````python
print("Please enter 'quit' when finished")
with open('guest_book.txt', 'a', encoding="utf8") as f:
    while True:
        name = input("Please write your name: ")
        if name == 'quit':
            break
        f.write(name + "\n")
        print(f"Greetings {name}, you are now in the guest book")
````

# Task 2: Fantasy Game Inventory
You are creating a fantasy video game. The data structure to model the playerʼs inventory will be a dictionary where:

* the keys are string values describing the item in the inventory
* the value is an integer value detailing how many of that item the player has.

For example, the dictionary value `{'rope': 1, 'torch': 6, 'gold coin': 42, 'dagger': 1, 'arrow': 12}` means the player has 1 rope, 6 torches, 42 gold coins, and so on.

Write a function named `display_inventory` that would take any possible `"inventory"` and display it like
the following:

Say you have a list value like this:
```
Inventory:
12 arrow
42 gold coin
1 rope
6 torch
1 dagger
Total number of items: 62
```

__Note!__ that the ordering of the items is not important. Some help can be found at the original project
description at [https://automatetheboringstuff.com/chapter5/](https://automatetheboringstuff.com/chapter5/).

__<font color='red'>If you want to have a challenge,</font>__
* sort the items alphabetically
* align them nicely such that the item name starts at the exact same position (and is not left-aligned like in the example)
* add singular/plural ("12 arrows"/"42 gold coins"), just adding 's' at the end is fine.

__suggested solution for regular task__

````python
inventory = {'rope': 1, 'torch': 6, 'gold coin': 42, 'dagger': 1, 'arrow': 12}
def display_inventory(inv):
    print("Inventory: ")
    total = 0
    for k,v in inv.items():
        print("{} {}".format(v,k))
        total = total + v
    print("Total number of items: {}".format(total))

display_inventory(inventory)
````

__suggested solution for challenge with sort and lambda__

````python
inventory = {'rope': 1, 'torch': 6, 'gold coin': 42, 'dagger': 1, 'arrow': 12}
def display_inventory(inv):
    print("Inventory: ")
    total = 0
    
    # remove lambda to sort on key instead of value
    for k, v in sorted(inv.items(), key = lambda x: x[1]): # lambda function is a small, anonymous function you write in a single line 
        # we'll talk about them later in the course
        print("{:02d} {}".format(v,k)) # :02d formats integers with to have at least two decimals, using zeros if value is below 10
        total = total + v
    print("Total number of items: {}".format(total))

display_inventory(inventory)
````

# Task 4: Analyzing a file
The file `sales.txt` file contains every element sold by a grocery store within the last 24 hours. Write a
program that prints the item that has been sold most frequently.

_Suggested Approach_: Read the file line-by-line and count frequencies using a dictionary. Find the item with
maximum frequency and return it (_hint_: use the `max()` function).

__suggested solution__

````python
sold = {}
with open('sales.txt') as f:
    for line in f:
        line = line.strip() # remove unnecessary '\n' charracter from line
        sold.setdefault(line, 0)
        sold[line] += 1

# one way to get the max from a dictionary is to use the key argumetn in the max function - however we have not discussed this yet
print(max(sold, key=sold.get))
# another solution is to do this
v = max(sold.values()) # find value with max value
i = list(sold.values()).index(v) # find the index for the value - here we need to transform the sold.values() into a list
k = list(sold.keys())[4] # find the key
print(k)
````

# Task 5: A more difficult analysis task
At the beginning of 2023, Sophia had 1500 kr on her bank account. In the file `bank.txt` you find all her
expenses and intake for the year 2023. Unfortunately, it's all scrambled and difficult for her to get an
overview over her finances. Each line in the file contains of two values separated by ",". The first is a day
(from 0 to 364, where 0 is January 1, ..., and 364 is December 31), the second is an integer that is either
positive (she deposited money) or negative (she withdrew money).

Solve the following subtasks:
1. Read the file and create a dictionary `bank_year` where keys are days, and values are a list of all bank
transactions on that day.
2. Starting from `bank_year`, create a dictionary bank_balance that contains the balance on Sophia's
account at the end of each day. (The initial balance of 1500 kr, plus all past transactions. For
example, if the transactions on the first day are [100,-30], then the balance at the end of this day will
be 1500 + 100 - 30 = 1570.)
3. Carry out the following analysis tasks:
    * on how many days were there no transactions at all?
    * on how many days is Sophia's account balance more than `200,000` kr?
    * What is her account balance on the evening of December 31?
    * How much did she __withdraw__ from her account over the entire year?
    * How much did she __deposit__ to her account over the entire year?

__suggested solution__

````python
# your code here
def read_file(fn):
    bank_year = {}
    with open(fn) as f:
        for line in f:
            day, amount = line.strip().split(", ")
            day = int(day)
            amount = int(amount)
            if day not in bank_year:
                bank_year[day] = []
            bank_year[day].append(amount)
    return bank_year

# load the file
bank_year = read_file("bank.txt")

balance = 1500
bank_balance = {}

# go through the file
for day in range(365):
    if day in bank_year:
        end_of_day_balance = 0
        for amount in bank_year[day]:
            end_of_day_balance += amount
        
        balance += end_of_day_balance

    bank_balance[day] = balance
````

````python
# How many days without transactions?
count = 0
for day in range(365):
    if day not in bank_year:
        count += 1
print("{} days had no transaction".format(count) )
````

````python
# how many days with more than 200,000 in account
count = 0
for day, balance in bank_balance.items():
    if balance > 200000:
        count += 1
print("{} days with balance > 200.000 kr.".format(count))
````

````python
# end of year balance
print("{} kr on December 31.".format(bank_balance[364]))
````

````python
# How much did she withdraw and deposit from her account over the entire year?
withdraw_count = 0
deposit_count = 0
for transactions in bank_year.values():
    for transaction in transactions:
        if transaction > 0:
            deposit_count += transaction
        else:
            withdraw_count -= transaction

print("She deposited {} kr, and withdrew {} kr.".format(deposit_count, withdraw_count))
````
