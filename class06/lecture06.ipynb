# Recap - methods, lists, and string operations
We talked about a lot... __seven most important things to remember__:

# Thing 1
### - a `list` is defined using either `[a,b,c]` or `list([a,b,c])`
### - a `tuple` is defined using either `(a,b,c)` or `tuple((a,b,c))`
### - a `set` is defined using either `{a, b, c}`, or `set([a,b,c])`

# Thing 2
### - lists contain ordered sequences, and are mutable (meaning you can change/update/delete the individual elements)
### - tuples contain ordered sequences, and are immutable (meaning you cannot change individual elements)
### - sets are not ordered, but containt unique values. They are mutable.

# Thing 3
### you can index or slice lists and tuples
### - inded by using `mylist[i]` or `mytuple[i]` - this extracts one value from a list or tuple
### - slice by using `mylist[i:j]` or `mytuple[i:j]` - this extracts a slice (if you slice a list it returns a list, if you slice a tuple it returns a tuple)

# Thing 4 - iterating over a list/tuple

### Last but not least, `.format()` a very powerful tool for formatting strings

````python
# so far we have inserted things into string in a very silly way
x = 3
print('Today I have eaten ' + str(x) + ' ice creams, and you have eaten ' + str(4) )
````

````python
# .format() is a different way of inserting variable values into literal text

# by uorder:
print('Today I have eaten {} ice creams, and you have eaten {}'.format('3', '4')) # by order (with empty curly braces)

# by position:
print('Today I have eaten {1} ice creams, and you have eaten {0}'.format('4', '3'))

# by variable name:
print('Today I have eaten {x} ice creams, and you have eaten {y}'.format(y = '4', x = '3')) # by variable name
````

## Reading a file in python - the good and the bad option
## - Bad option - reading the whole file
## - Good option - reading the file line by line

## To work with files you need to:
#### - Open a file (using the built-in function `open()`)
#### - Read a file (using a method `.read()` )
#### - Close the file (can use the method `.close()`)

## Lets open the file 'overview.txt' (the bad option which loads the whole file in memory)

````python
# this example is to showcase the open function - don't do this at home
f = open('overview.txt')
f # it returns a write - nothing has been read yet
````

## To read the file you need to read its contents

````python
lines = f.read()
print(lines) # printing its contents
````

## and then you have to remember to close the file using the `.close()` method

````python
f.close()
````

# The good way (reading line by line, i.e. not loading everything in memory)
### The `with` statement: simplifies resource management by automatically handling setup and cleanup, ensuring files or connections close safely even if errors occur. 

### The `as` keyword: used to create an alias for a module, object, exception, or file. It helps make code shorter, more readable, and easier to work with.

````python
with open('overview.txt') as f:
    for line in f: # reads line by line
        print(line)
````

### You can also read the whole file using `with` - but there is no reason to do so

````python
with open('overview.txt') as f:
    contents = f.read()
print(contents)
````

## Always use the `with` statement as it automatically closes files! 

## There are different ways of reading
I mostly use this variant:
```python
with open('overview.txt') as f:
    for line in f:
        print(line)
```
Because you often want to process each line in a certain way.

However, you can also use the `.readlines()` method (this still loads the whole file in memory):
```python
with open('overview.txt') as f:
    content = f.readlines()
```

````python
with open('overview.txt') as f:
    content = f.readlines()
print(content) # individual lines are now elements in a list
````

# Writing to a file
### the two modes of writing
### 1️⃣ writing a new file (`"w"`)
### 2️⃣ appending to an existing file (`"a"`)
### difference through how file is opened (<font color='red'> append does not overwrite a file </font>)


## You always write string(s) to a file
### Lets create a new file using information stored in a list
```python
grades_to_save = [['Sandra',10],['Bob',7],['Alice',12]]
```

````python
grades_to_save = [['Sandra',10],['Bob',7],['Alice',12]]

with open('grades.txt','w') as file:
    for name, grade in grades_to_save:
        file.write(name + ': ' + str(grade))
````

````python
# lets quickly see the contents of the file
!head grades.txt # what happened there?!
````

### You need to explicitly add newlines

````python
grades_to_save = [['Sandra',10],['Bob',7],['Alice',12]]

with open('grades.txt','w') as file:
    for name, grade in grades_to_save:
        file.write(name + ': ' + str(grade) + '\n') # <--------- added newline using '\n'
````

````python
# lets quickly see the contents again
!head grades.txt # much better
````

## Remember the `.format()` method for strings! ⭐️

````python
grades_to_save = [['Sandra',10],['Bob',7],['Alice',12]]

with open('grades.txt','w') as file:
    for name, grade in grades_to_save:
        file.write('{}: {}\n'.format(name,grade)) # much easier to write
````

````python
!head grades.txt
````

# Exceptions - a really powerful tool
### `exception` statements allow you to react on behavior that would otherwise result in a crashed program

```python
try:
    code_block
except:
    code_block_if_error_occured
```

# simple example

````python
print(4/0) # this thrown an error because we cant divide by zero
````

````python
try:
    print(4/0)
except:
    print('Uh oh, dividing by zero is undefined')
````

# When to use exceptions?
### Exceptions are used if the program uses resources outside of our control, e.g.,
#### - user input that could fail
#### - files that may not exist
#### - In the except block, we can

### Exceptions specify the error that should be “caught"

## By default except will catch all errors - not always useful

````python
try:
    print(4/0)
except ZeroDivisionError: # only catches the zero division error
#except NameError: # this does not catch the error
    print('Uh oh, dividing by zero is undefined')
````

## another example

````python
try:
    with open('file-who-does-not-exist.txt') as f:
        for line in f:
            print(line)
except FileNotFoundError:
    print('File does not exist, loading a random file instea')
    # something else
````

# When not to use exceptions
### In general, specify what exactly the error, don’t just use `except:`
### make sure it’s the correct one!
### don’t use try/except as a means to make sure that your program doesn’t crash

# Today's fun part - dictionaries 📕
### A dictionary is a "collection of key-value pairs, with unique keys"

```python
{key1:value, key2:value, key3: value}
```

### each key is unique and associate with a value

# How can we store pairs of data?
```python
# example of a dictionary
my_cat = {
'age': 12,
'name': 'Santa Claws',
'color': 'grey',
'size': 'large' }
```

### a dictionary is defined using `{ }`  (<font color='red'>this is different from sets!</font>)
### contains sequence of key : value pairs, separated by commas
### key and values must both be Python values (`int`, `float`, `str`, `bool`, etc.)

## can also be created with `dict()` function

````python
{'name': 'Santa Claws', 'color': 'grey'} == dict({'name': 'Santa Claws', 'color': 'grey'})
````

# Accessing Values in a Dictionary
### Access values associated with `key` using `my_cat[key]`, where `key` is one of the keys in the dictionary

````python
my_cat = {'age': 12, 'name': 'Santa Claws', 'color': 'grey', 'size': 'large' }

# Here keys: 'age', 'name', 'color', 'size' are all strings
my_cat['name']
````

### accessing a key which does not exist

````python
my_cat = {'age': 12, 'name': 'Santa Claws', 'color': 'grey', 'size': 'large' }

my_cat['danger-level']
````

## Lists vs dictionaries
### List: Lists are ordered - there is a first, second, third element
### _Dict_: Dictionaries are unordered - there is no first, no second, no third element

## Lists vs dictionaries
### List: Access list elements via an index
### _Dict_: Access dict values/elements via a key

## Lists vs dictionaries
### List: can produces slices, i.e., sublists
### _Dict_: no slices possible
<br/>

### List: order matters
### _Dict_: order is not interesting

````python
[0, 200] == [200, 0]
````

````python
{'age':20, 'birthday': 'no'} == {'birthday': 'no', 'age':20}
````

# two more dictionary methods 🫠
### `.get()` 
### `.setdefault()`

### they allow to write somewhat nicer code, but they can always be replaced by more fundamental methods. __Try to use them, but if you feel uncomfortable use the basic functionality__

# The `.get()` method
### `get(key, default)` provides a safe way of accessing a dictionary, returning a default value if a key is not present
   

````python
# simple example
my_cat = {'age': 12, 'name': 'Santa Claws', 'color': 'grey', 'size': 'large' }
my_cat.get('nationality','Your cat has no nationality')
````

````python
# more complex example
ice_creams_eaten = {'Alice': 3, 'Bob': 1}
for name in ['Alice', 'John', 'Bob', 'Meredith']:
    print('{} has eaten {} icecreams'.format(name,ice_creams_eaten.get(name,0)))
````

# the `.setdefault()` method
### returns the value of the specified key. If the key does not exist, insert it as a new key, with the specified value

````python
# a silly example
my_cat = {'age': 12, 'name': 'Santa Claws', 'color': 'grey', 'size': 'large' }
print(my_cat)
my_cat.setdefault('weight','fat')
print(my_cat)
````

````python
# more meaningful example
ice_cream_eaters = ['John','Sally','Bob','Bob','Bob','Alice','Bob','Sam','Angela','John']
ice_creams_eaten = {} # create an empty dict

for person in ice_cream_eaters:
    ice_creams_eaten.setdefault(person,0) # sets a default value to zero
    ice_creams_eaten[person] += 1

print(ice_creams_eaten)
````
