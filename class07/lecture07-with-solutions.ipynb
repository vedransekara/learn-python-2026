### the `setdefault` method
A handy method for setting default values while avoiding overwriting existing ones.
`setdefault`:

* Returns the value of a specified key, if it exists
* If the key does not exist, it adds the key with a default value and returns that default

````python
ice_eaters = ['Sarah', 'Mette', 'Sarah', 'Mette', 'Brian', 'Brian', 'Brian', 'Brian', 'Mette', 'Paul', 'Brian', 'Paul', 'Sarah',
 'Søren', 'Sarah', 'Brian', 'Paul', 'Paul', 'Brian', 'Mette', 'Brian', 'Brian', 'Søren', 'Brian', 'Marie', 'Brian', 'Mette',
 'Søren', 'Brian', 'Marie', 'Marie', 'Marie', 'Marie', 'Brian', 'Søren', 'Brian', 'Søren', 'Paul', 'Paul', 'Paul', 'Marie',
 'Sarah', 'Sarah', 'Sarah', 'Mette', 'Sarah', 'Søren', 'Sarah', 'Mette', 'Paul', 'Marie', 'Sarah', 'Brian', 'Brian', 'Mette',
 'Søren', 'Mette', 'Mette', 'Sarah', 'Brian', 'Sarah', 'Brian', 'Brian', 'Sarah', 'Sarah', 'Sarah', 'Brian', 'Mette']

# without setdefault
eater_count = {}
for eater in ice_eaters:
    if eater not in eater_count:
        eater_count[eater] = 0
    eater_count[eater] += 1
print(eater_count)
````

````python
# with setdefault
eater_count = {}
for eater in ice_eaters:
    eater_count.setdefault(eater,0) # set defaul, or do nothing
    eater_count[eater] += 1
print(eater_count)
````

### now lets calculate how many ice creams 🍦 people eat on average

````python
total_icecreams = 0
people = 0
for person, count in eater_count.items():
    people += 1
    total_icecreams += count
print(f'Average number of ice creams a person eats {total_icecreams/people}')
````

## Regular expressions (`re`) module

https://docs.python.org/3/howto/regex.html

___Regular expresions are...___

* "**tiny, highly specialized programming language** where you can specify the rules for the set of possible **strings that you want to match**"
* a well-known concept in computer science
* used by many programming languages (perl, java, python...)
* all about **finding matching string patterns**
* useless to learn by heart (imho) - just know that they **exist**, in case you might need them

```python
# to use the re module, import it first:
import re
```

### A gentle intro to `re.search()` and `re.findall()`
```python
# string: the string we're searching within
# pattern: the pattern we're trying to find
# it can be LITERAL (explicitly stating the pattern), or 
# expressed with REGEX - see list below
re.search(pattern, string) # will show you whether pattern is in the string
re.findall(pattern, string) # will return all matching patterns from the string

```

Patterns has to be formatted in a special way to get it to work with regex:
```
pattern = r'some_patterns'
```
__Notice the `r` before the string! It tells python to interpret this as a "raw" string, meaning you can freely use `\` (backslashes)__

**How to make your own regex patterns (some EXAMPLES)**
* `\d` Matches any number
* `\D` Matches any non-number
* `\w` Matches any whitespace
* `\W` Matches any non-whitespace
* `+` matches the preceding character 1 or more times
* `[]` specifies a class you want to match
* `-` within `[]` indicates a range
* `[a-z]` matches all small letters from a to z

````python
# first, we need to import the re module
import re
# now we can use it in this notebook
````

````python
# re.search(pattern, string) tells us whether a pattern is contained in a string 
mypattern = 'a' # try out other values
mystring = 'The Fast and The Furious 11'
re.search(mypattern, mystring)
# if yes: we get back a "Match object" (don't worry about it); 
# if no: we get back None (nothing)
````

````python
# re.findall(pattern, string) RETURNS all pattern matches in a string 
mypattern = 'a' # try out other values
mystring = 'The Fast and The Furious 11'
re.findall(mypattern, mystring)
````

````python
# find only the numbers in the string: "\d" pattern
mypattern = r'\d' # this means: a single character which is a number
mystring = 'The Fast and The Furious 11'
re.findall(mypattern, mystring) # we found the numbers 1 in the string, but they actually should belong together
````

````python
# find only the numbers in the string: "\d" pattern
mypattern = r'\d+' # this means: ONE OR MORE CONSECUTIVE characters which are number
mystring = 'The Fast and The Furious 11'
re.findall(mypattern, mystring) # we found the numbers 1 and 1 in the string, but they actually should belong together
````

````python
# find everything which is NOT a number, 
mypattern = r'\D' # this means: every single character that is NOT a number
mystring = 'The Fast and The Furious 11'
re.findall(mypattern, mystring)
````

````python
# find everything which is NOT a number, 
mypattern = r'\D+' # this means: ONE OR MORE CONSECUTIVE characters that are NOT numbers
mystring = 'The Fast and The Furious 11'
re.findall(mypattern, mystring)
````

````python
# specify a class within a certain range
mypattern = r'[A-Z]' # all single uppercase letters
mystring = 'The Fast and The Furious 11'
re.findall(mypattern, mystring)
````

# List comprehension

A powerful, "Pythonic" way to create lists (and to replace for-loops).
```
[expression for item in iterable]
```
__Outputs a list!__

````python
# e.g. finding the square of all numbers form 0 to 10
[i**2 for i in range(0,11)]
````

Here:
* __expression:__ `i**2`
* __item:__ `i`
* __iterable:__ `range(0,11)`

### List comprehensions always output a list

````python
# you can assign the list comprehension to a variable
new_list = [i**2+3 for i in range(0,11)]
new_list
````

### you can even use variables in the list comprehension

````python
# you can assign the list comprehension to a variable
new_list = [i**2+3 for i in range(0,11)] # the list comprehension from before
new_new_list = [j/2+1.5 for j in new_list] # using it here again
new_new_list
````

### I often use list comprehensions to unpack lists of lists, or lists of tuples

````python
ice_creams_eaten = {'Marie': 7, 'Sarah': 15, 'Paul': 8, 'Brian': 20, 'Mette': 11, 'Søren':7}
[(v,k) for k,v in ice_creams_eaten.items()]
````

### why even do this?
#### Well...in python there is are `max`/`min` functions to find the largest/smalles item in a list

````python
ice_creams_eaten = {'Marie': 7, 'Sarah': 15, 'Paul': 8, 'Brian': 20, 'Mette': 11, 'Søren':7}
max(ice_creams_eaten.items()) # if you do this it finds the max of the first item
````

__list comprehensions easily allow you to find the max of the other__
#### (Psst....There is also another approach to do this a bit easier (we'll get to it later today))

````python
ice_creams_eaten = {'Marie': 7, 'Sarah': 15, 'Paul': 8, 'Brian': 20, 'Mette': 11, 'Søren':7}
max([(v,k) for k,v in ice_creams_eaten.items()]) # it finds the max of the first item again, but now we have switched them around
````

### you can even include `if` statements in list comprehensions
```
[expression for item in iterable if condition]
```

````python
even_numbers = []
for i in range(21):
    if i % 2 == 0: 
        even_numbers.append(i)
even_numbers
````

````python
# adding a condition in a list comprehension
[i for i in range(21) if i % 2 == 0]
````

````python
# ... and as usual, it works for all other iterables, e.g. strings
# create a list of all letters in a word that are NOT vowels
word = "ambivalence"
[letter for letter in word if letter not in ['a', 'e', 'i', 'o', 'u']]
````

# `lambda` functions (anonymous functions)
### Sometimes a function is so simple (and has such a specific use case) that you do not want to give it a name

###  using the `lambda` keyword python allows you to create simple functions in a single line 

# example
```
def smaller_than_ten(x):
    return x < 10
```

__you can use a lambda function__
```
lambda x: x < 10
```

<font color='red'> __notice, `lambda` functions do not need a `return` statement__ </font>

# `lambda` function structure

```
lambda arguments: expression
```
__lets go back to our ice cream example__

````python
ice_creams_eaten = {'Marie': 7, 'Sarah': 15, 'Paul': 8, 'Brian': 20, 'Mette': 11, 'Søren':7}
sorted_ice_cream_eaters = sorted(ice_creams_eaten.items(), key=lambda x: x[1])
print(sorted_ice_cream_eaters)
````

### another example - sorting words according to length

````python
words = ['apple', 'pie', 'banana', 'cherry','pineapple','tomato','rhinoceros']
sorted_words = sorted(words, key=lambda x: len(x))
print(sorted_words)
````

### or finding the max in a list of tuples/list

````python
ice_creams_eaten = {'Marie': 7, 'Sarah': 15, 'Paul': 8, 'Brian': 20, 'Mette': 11, 'Søren':7}
max([(v,k) for k,v in ice_creams_eaten.items()]) # it finds the max of the first item again, but now we have switched them around
````

````python
max(ice_creams_eaten, key=lambda x: ice_creams_eaten[x]) # using a lambda function
````

### lambda functions are really powerfull especially combined with the built in function `map`
__the `map()` function applies a function to every element of one (or more) iterable(s) and returns a map object (iterator) containing the transformed results. It is often used for element-wise operations where it replaces explicit loops with shorter and more readable code__

```
map(function, iterable)
```

````python
some_strings = ['1', '2', '3', '4', '5', '6', '7', '8']
list(map(lambda x: int(x), some_strings))
````

### Or transforming temperature data from Farenheit to Celcius

````python
farenheit = [32, 68, 98.6, 212]
celcius = list(map(lambda t: (t-32)*(5/9),farenheit))
print(celcius)
````
