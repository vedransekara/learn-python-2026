## Faster computation of the sum

````python
def sum_to_n_gauss(n):
    return (n+1)*n//2 # sum formula of Gauss
````

````python
%timeit sum_to_n_gauss(100)
````

````python
def quadratic(n):
    for i in range(n):
        for j in range(n):
            print('*', end='')
        print()
quadratic(4)
````

````python
def cubiq(n):
    for i in range(n):
        for j in range(n):
            for k in range(n):
                print('*', end='')
            print()
        print()
cubiq(4)
````

## Search
Gets two inputs, an element `x` and a list `L` and its supposed to say whether `x` is in `L`

<font color='red'>__what is the running time of this algorithm?__</font>

````python
def search(x, L):
    """True if x an element of L."""
    for y in L:
        if x == y:
            return True
    return False
````

````python
%%timeit search(-1,range(1000))
````

## There are some things we need to discuss here:
* Running time of $O(n)$, where $n$ is the length of the list
* Often we want to estimate running times for the worst case scenarios
* Size of numbers in the list dont have an impact on running times, i.e. `search(-1,range(100))` ~ `search(-1,range(100,200))`
* `x` can have an impact on the running time 

> In the example above, we benchmarked the function by searching for an element that is not part of the list. This is the worst-case scenario: If the element had been in the collection, it can stop as soon as it was identified. In running time analysis, we care about the worst-case behavior. The goal is to obtain an upper bound on the running time that is true no matter how unlucky you are with the presented input.

## Lets look at a different implementation of Search
Here we are using the `in` operator form python to test if an element is part of a list

<font color='red'>__what is the running time of this algorithm?__</font>

````python
def search(x, L):
    """True if x is an element of the collection L."""
    return x in L 
````

````python
%%timeit search(-1,range(1000))
````

* Difficult to estimate runtime when using built-in functions or functions imported from external modules
* Try to compare run time for sets and lists
* Running time depends on the choice of data structure for `L`. If `L` is a list, running time $O(n)$, if `L` is a set, then the running time is $O(1)$.

## Arithmetic
Given an number `1234` the below function take the sum of the individual digits, i.e. `1 + 2 + 3 + 4 = 10`

<font color='red'>__what is the running time of this algorithm?__</font>

````python
def sum_of_digits(n):
    """Returns the sum of digits in an non-negative integer n."""
    string_n = str(n)
    res = 0
    for c in string_n:
        res += int(c)
    return res
````

````python
%%timeit sum_of_digits(1234567890)
````

* Running time is linear in the number of digits.
* __Is there a mathematical function which, given an integer, gives you the number of digits?__

__Solution: The number of digits is the logarithm of n to base 10, $\log_{10}(n)$ (the x such that 10^x = n).__
