Alice regularly travels between Denmark and Germany. Since both countries use different currencies,
Alice always needs to convert values between the two currencies.

To help Alice, define two functions `eur_to_dkk(amount)` and `dkk_to_eur(amount)` that take as

### Coding task 4 (The Collatz sequence)
Write a function named `collatz` that has one parameter named `number`. If number is even, then `collatz`
should print `number // 2` and return this value. If number is odd, then `collatz` should print and return `3 * number + 1`.

Then write a program that lets the user type in an integer (using the `input` function) and that keeps calling `collatz` on that number
until the function returns the value 1. (Amazingly enough, this sequence actually works for any integer —
sooner or later, using this sequence, youʼll arrive at 1!)

__Hint:__ An integer number is even if `number % 2 == 0`, and itʼs odd if `number % 2 == 1`.

The output of this program could look something like this:
```
Enter number:
3
10
5
16
8
4
2
1
```

````python
# YOUR CODE HERE
def collatz(number):
    if number % 2 == 0:
        return number//2
    elif number % 2 == 1:    
        return 3 * number + 1

print('Enter number:')
number = int(input())

if type(number) == int:
    while number != 1:
        number = collatz(number)
        print(number)
else:
    print('you must be tripping this is not an integer')
````

### Coding task 5
We have the task to write a piece of software for PostNord. The software should read three pieces of
information from the user: The street name, the zip code, the city.

Write a function `get_address(street_name, zip_code, city)` that returns a string with the
following format:

* If the user enters all information, the string should be formatted as in the following example: `Retortvej, 2500, Valby`

* However, sometimes we are not sure about the zip code of the city. If the user doesn't enter a zip
code, the format should be `Retortvej, Valby` (and not `Retortvej,, Valby`).

That means that the program below should produce the following output:

```
Please enter the street name: Amagerbrogade
Please enter the zip code: 2300
Please enter the city: Copenhagen
Amagerbrogade, 2300, Copenhagen
```

And

```
Please enter the street name: Amagerbrogade
Please enter the zip code:
Please enter the city: Copenhagen
Amagerbrogade, Copenhagen
```

Your program starts as follows:

```python
street = input("Please enter the street name: ")
zip_code = input("Please enter the zip code: ")
city = input("Please enter the city: ")
# define your function here
# the following should work
print(get_address(street, zip_code, city))
```

````python
# YOUR CODE HERE
street = input("Please enter the street name: ")
zip_code = input("Please enter the zip code: ")
city = input("Please enter the city: ")

def get_address(street, zip_code, city):
    if zip_code != "":
        return street + ", " + zip_code + ", " + city
    else:
        return street + ", " + city

print(get_address(street, zip_code, city))
````

__Extra challenge__: Take a look at [https://en.wikipedia.org/wiki/Postal_codes_in_Denmark](https://en.wikipedia.org/wiki/Postal_codes_in_Denmark). For each zip_code, append the region in Denmark to send the mail to. Only split up into regions "Zealand", "Funen",
"Bornholm", and "Jutland". This functionality should be provided by defining a function `get_region(zip_code)` that returns the region to which `zip_code` belongs to.

````python
# YOUR CODE HERE
def get_region(zip_code):
    if (1000 <= zip_code < 3700) or (4000 <= zip_code < 5000):
        return 'Zealand'
    elif 3700 <= zip_code < 3800:
        return 'Bornholm'
    elif 5000 <= zip_code < 6000: 
        return 'Funen'
    elif 6000 <= zip_code < 100000:
        return 'Jutland'
    else:
        'Unknown'
````

````python
# YOUR CODE HERE
street = input("Please enter the street name: ")
zip_code = input("Please enter the zip code: ")
city = input("Please enter the city: ")

def get_address(street, zip_code, city):
    # get region first
    region = get_region(int(zip_code))
    
    if zip_code != "":
        return street + ", " + zip_code + ", " + city + ", " + region
    else:
        return street + ", " + city

print(get_address(street, zip_code, city))
````
