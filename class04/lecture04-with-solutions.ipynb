## Brief recap of `for` loops
### for loops repeat a block of code a fixed amount of time

````python
# range() starts at 0 by default; 
for i in range(3,7):
    print(i)

# this is interpreted as range(start, stop), start is optional
````

````python
# range() counts +1 at each step by default;
# we can tell range to use a different step size:
for i in range(5,125,10):
    print(i)

# all options for range are range(start, stop, step), start, and step are both optional
````

````python
# we have already tried iterating over (using the for loop)
# for lists, and range objects. we can do the same for strings:
for letter in "Happy birthday":
    print(letter)
````

````python
# what you can NOT iterate over:
for number in 5: # ints, floats, booleans
     print(number) # this throws an error message. 
# those are usually also the objects that have NO LENGTH!
# len(5)
````

## brief recap of `while` loops

### makes a block of code execute over and over again, as long as a condition is true (looks similar to an `if` statement)

````python
# how to move on from the while loop?!
# option 1: the condition must change INSIDE the while loop
i = 0
while i < 5: 
    print("i is: ", i)
    i = i + 1 # at each step, we CHANGE the condition that is evaluated
````

````python
# how to move on from the while loop?!
# option 2 - use the break statement
i = 0
while True: 
    print("i is: ", i)
    i = i + 1 # at each step, we CHANGE the condition that is evaluated
    if i >= 5:
        break
````

# Beware - there is a function who shall not be named or used - `global`
* If you want to modify a variable eggs in global scope in a function, you have to use the statement `global eggs`

````python
def spam():
    eggs = 'spam eggs'

eggs = 'global eggs'
spam()
print(eggs)
````

````python
# example of the use of the global
def spam():
    global eggs
    eggs = 'spam eggs'

eggs = 'global eggs'
spam()
print(eggs)
````

# Helping out the users: doc strings
### Functions are supposed to be used as "blackboxes"
* We don’t know how `print` works internally, we know what to expect from it
* We don’t know how `len` works internally, we know what to expect from it

### Its important to describe to users what to expect a function does
* __This is achieved using a "docstring"__

## A docstring provides a clear description of expected input/output behavior

<mark style="background-color: lightgreen"> __Programming best practice: A function should have an instruction manual, the docstring (documentation string). Once defined, you can access the docstring by ?function_name__<span>

````python
?print
````

### you can define doc strings using `"""some text"""` (written just below the definition of a function)

````python
# example
def complex_math(a,b,c):
    """Takes 3 numbers a, b, c as input and calculates a**b - c
    * a,b,c: can be integers or floats
    Returns: the calcualted value
    """
    
    return a**b - c
````

````python
?complex_math
````

# Keyword parameters
* Python allows parameters to have default values
* these are called keyword parameters
* these parameters don’t need to be provided (i.e. optional when calling a function)
* their name can be used when calling the function
* all keyword parameters must be at the end of the parameter list (Why?)

# defining functions with keyword parameters

````python
def hello(name, title = 'Doctor'):
    print('Hello ' + title + ' ' + name) 
````

````python
hello('Alice')
hello('Alice', 'Doctor') # be careful with this - it assumes function has only two parameters
hello('Alice', 'Professor') # be careful with this - it assumes function has only two parameters
hello('Alice', title = 'Professor')
````
