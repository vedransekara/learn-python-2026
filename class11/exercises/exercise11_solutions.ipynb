## Task 2a: Find the top 10 most oil producing countries across all years
Pandas lets you iterate over indices and rows, using
```python
for index, row in df.iterrows():
    do_something
```

This is a traditional _pythonic_ way of terating over data in a DataFrame. Its super __inefficient(!) and slow(!)___ (we will learn how to avoid this next lecture), but here you have to use it.

Use `iterrows` to iterate over rows and find the top 10 most oil producing countries.

````python
total_oil = dict()
for i, row in oil.iterrows():
    country = row['Country or Area']
    total_oil.setdefault(country,0)
    total_oil[country] += float(row['Quantity'])
````

````python
# here we take the items of the total oil dictionary, and sort it according to the quantity,
# using a lambda function to indicate which index we are sorting after, and using reverse=True we get descending values
sorted(total_oil.items(),key=lambda x: x[1],reverse=True)[:10]
````

````python
# we start by using a list comprehension to create a country to region dictionary 
country2region = dict([(countries[i],regions[i]) for i in range(len(countries))])
````

````python
# then we use the apply function to add this as a column to the oil dataframe
oil['region'] = oil['Country or Area'].apply(lambda x: country2region.get(x,'Other'))
````

````python
# to count up oil for each region you can take multiple approaches. 
# one obvious choice is to use the same method as 2.1.
# Below, however, i demonstrate a different way of doing it

# First we iterate over all unique regions
oil_per_region = []
for r in set(oil['region']):
    # filter dataframe on region and sum up all quantities
    total_oil_for_region = oil[oil['region'] == r]['Quantity'].sum()
    # append to list
    oil_per_region.append((r,total_oil_for_region))

# then sort the list
sorted(oil_per_region,key=lambda x: x[1],reverse=True)
````
