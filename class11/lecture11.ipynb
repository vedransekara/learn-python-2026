# Recap of numpy

```python
import numpy as np # traditionally imported as np, but you are free to use any acronym you like
```

### defining arrays 
* vectors = 1-dimensional arrays
* matrixes = 2-dimensional arrays
* tensors = 3+ dimensional arrays


```python
# defining a vector (1-dimensional array) form a python list
v = np.array([1,2,6,1,7,3,5,9,3,2,7,21,7,0])
```

```python
# defining a vector (1-dimensional array) form a python tuple
v = np.array((1,2,6,1,7,3,5,9,3,2,7,21,7,0))
```

```python
# defining a matrix (2-dimensional array) from a list of lists
M = np.array([[1,2,3,4],[2,4,6,8],[3,6,9,12],[4,8,12,16]])
M
```

```python
# to transform a dictionary to an array requires a bit more work
d = dict({1:'123',2:'asa',5:'13',9:'1231',999:False})

# this transforms it to a two-dimensional array, i.e. a matrix
np.array(list(d.items()))
```

__be careful with when you are working with nested dictionaries, as will also become part of the array!__

### Indexing and slicing
Its very similar to lists, but not exactly the same!

__1-dimensional arrays:__
  * indexing: `array[element]`, e.g. `v[2]`
  * slicing `array[start:stop]`, e.g. `v[0:5]`

```python
# for 1-dimensional arrays it selects items directly
v = np.array([1,2,6,1,7,3,5,9,3,2,7,21,7,0])
print(v[2]) # selecting one element
print(v[0:5]) # selecting a slice
```

__For 2-dimensional arrays:__
    * select an individual element `array[row_i, column_i]`, e.g. `M[0,2]`
    * select a slice of array `array[row_start:row_end, column_start:column_end]`, e.g. 

```python
# defining a matrix (2-dimensional array) from a list of lists
M = np.array([[1,2,3,4],[2,4,6,8],[3,6,9,12],[4,8,12,16]])
print(M)
print(M[1,1]) # selecting one element
#print(M[1:3,1:3])
```

### <font color='red'> Remember: An index slice only creates a view! </font>
__Data is not copied, and any modifications to the view will be reflected in the source array! 🫠__

```python
arr = np.arange(10) # defining an array of integers from 0 to 9 (including)
print(arr)
arr_part = arr[5:8]
arr_part
```

```python
arr_part[:] = 666
arr
```

## you can easily do math with arrays

```python
# operations on the whole array
M = np.array([[1,2,3,4],[2,4,6,8],[3,6,9,12],[4,8,12,16]])
M * 2/3 - 1
```

```python
v = np.array([1,2,6,1])
# multiplication of vector with metrix
v*M # remember the dimensions have to match
```

```python
# dot products 
v1 = np.array([1,2,6,1,3,6])
v2 = np.array([0,9,5,1,2,-2])
np.dot(v1,v2)
```

```python
# length of arrays
print(np.linalg.norm(v1))

# same as  
print(np.sqrt(sum(v1**2)))
```

## Masking
For NumPy:
* `&` means `and`
* `|` means `or`

```python
M = np.array([[1,2,3,4],[2,4,6,8],[3,6,9,12],[4,8,12,16]])

mask = (M > 2) & (M < 5)
M[mask]
```

# Subsetting DataFrames

* Boolean indexing (filtering by conditions)
* bitwise operators!

# Boolean indexing (by condition)

##### `df[columnlabel]` can be combined with comparison `> < == !=` operators
##### `df[condition]` returns only those rows where condition is True
##### `df[(condition1) & (condition2)]` returns only those rows where both conditions are True
##### `df[(condition1) | (condition2)]` returns only those rows where either conditions are True

```python
# let's see what happens if we use a comparison operator with a single column:
# returns for each row either True or False
df["Age"] > 18 # was this person over 18 on that ship?
```

```python
# we can use that condition to index only rows where condition is True:
my_condition = df["Age"] > 18 # who was over 18 on that ship?
df[my_condition] 
```

```python
# a shorter (but perhaps more confusing at first) way to write this: 
# df[condition] (where condition contains a df column)
df[df["Age"]>18]
```

```python
# filtering by several conditions:
# put each condition inside () round brackets
# combine them with & (meaning "and") or | (meaining "or")
# everyone that was over 18 AND survived 
df[(df["Age"]>18) & (df["Survived"]==0)]
```

# Combining several dataframes
### `pd.concat([<list-of-dataframes>],axis=0/1)`

### concatenating dataframes along `axis=0` (adding new rows)

```python
# lets make two artificial datasets
adults = df[df.Age >= 18].copy() # using .copy() creates new dataframes
kids = df[df.Age < 18].copy() 
# merge datasets again
merged_again = pd.concat([adults,kids], axis=0) # axis=0 means along rows
```

### concatenating along `axis=1` (adding new column)

```python
# again lets make two artificial datasets
names = df["Name"].copy() # take out name column
rest = df.drop(labels="Name", axis = 1).copy() # drop it from dataframe
# add column back using concat
pd.concat([names,rest], axis=1)
```

# Filtering out missing values (NaNs) with boolean conditions

* `NaN` .... "not a number"
* Pandas functions: `isna()` and `notna()` (`isna` = is not available, `notna` = not not available)
* Note, for pandas: "not available" (covers both `NaN` and `None`)

```python
df['Cabin'].head()
# for rows with index 0, 2, and 4 information on Cabin is missing
```

```python
df["Cabin"].isna()
# returns for each row whether data is NOT AVAILABLE (True) or available (False)
```

```python
df["Cabin"].notna()
# returns for each row whether data is AVAILABLE (True) and not available (False)
```

### How to use `isna` and `notna` to clean data?

```python
# boolean indexing to only have rows where we know the Cabin value:
my_condition = df["Cabin"].notna()
df[my_condition]
# or, shorter: df[df["Cabin"].notna()]
```

This only removes `NaN`'s in the Cabin column.

But, there can still be `NaN`'s in other columns

```python
df[(df['Age'].notna()) & (df['Cabin'].notna())]
# remember, this creates a vies, you need to overwrite variable to make it permanent
```

## Tilde operator `~` in pandas means `not`

So `df[~df[column].isna()]` is equivalent to `df[df[column].notna()]`

```python
df[~df['Age'].isna()]
# this and the below expression return the same
```

```python
df[df['Age'].notna()]
```

## Saving DataFrames to file
