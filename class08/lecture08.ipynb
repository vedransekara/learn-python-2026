## Recap - list comprehensions
### A powerful, _"Pythonic"_ way to create lists - without using the '_typical_' loop structure

## List comprehensions come in three forms

* Regular for loop structure `[expression for item in iterable]` where you just want to iterate over something
* Loop with a codition `[expression for item in iterable if condition]` where you also want to use a if statement/condition
* Loop with if/else structure `[expression1 if condition else expression2 for item in iterable]` were you want multipe options depending on condition

### List comprehensions are really neat to use when you e.g .want to filer and transform values at the same time

`````python
# for example, here we want to square all even numbers - using a for loop 
numbers = [1, 2, 3, 4, 5, 6]
squares_of_even_numbers = []

for n in numbers:
    if n % 2 == 0:
        squares_of_even_numbers.append(n ** 2)

print(squares_of_even_numbers)
`````

`````python
numbers = [1, 2, 3, 4, 5, 6]
squares_of_even_numbers = [n**2 for n in numbers if n%2 == 0]
print(squares_of_even_numbers)
`````

### What is the benefit here of a list comprehension?

```python
squares_of_even_numbers = [n**2 for n in numbers if n%2 == 0]
```

* Its shorter: 1 line vs 4 lines
* You don't need to create an empty list first
* You don't need to append to the list

## Top tips with list comprehensions
### ✅ Use a list comprehension when you're building a new list from an existing iterable.
### ❌ Use a regular for loop when the logic is complex, involves multiple steps, error handling, or side effects.

# the `requests` library

### `requests` is a Python package to send HTTP requests and receive HTTP responses, fortunately it comes pre installed with Anaconda 👍 (https://docs.python-requests.org/en/latest/index.html) </font>
### <font color='dcdcdc'> Everytime you visit a page on the internet, your browser submits a request to a server, retreives the content, and shows it to you using after its been processed </font>



`````python
import requests
`````

## `requests` has one important method/function `.get()`

`````python
# the function .get(url) sends a request; 
# and returns a response
my_url = 'https://en.itu.dk/Programmes/BSc-Programmes'
requests.get(my_url)
`````

Requests will print a HTML response code. In this case 200 means, '_OK_' a standard response for a successful HTTP request.

You have problably at some time encountered HTML code 404, which means page 'not found'.

You can find a full list of HTML codes [here](https://en.wikipedia.org/wiki/List_of_HTTP_status_codes.

### We want to save the request to a variable

`````python
# # the function .get(url) sends a request; 
# and returns a response
my_url = 'https://en.itu.dk/Programmes/BSc-Programmes'
my_response = requests.get(my_url)
`````

`````python
# my_response us it type Response :-|
type(my_response)
`````

### a Response type has several attribtes

`````python
# ... and has several attributes, for example:
print("--> status code:", my_response.status_code)
print("--> encoding attribute:", my_response.encoding)
print("--> url attribute:", my_response.url)
print("--> text attribute:", my_response.content) # the is one of the most important 
`````

# How the hell do we process HTML text? 🤷
(You can learn more about HTML [here](https://www.w3schools.com/html/))
### Solution: the `beautifulsoup4` package - it should also be pre-installed with the anaconda package
Beautiful soup homepage --> https://www.crummy.com/software/BeautifulSoup/

(According to wikipedia) It takes its name from the poem Beautiful Soup from Alice's Adventures in Wonderland and is a reference to the term "tag soup" meaning poorly-structured HTML code.

### We will use beautifulsoup4 to "parse" (read) the content of our HTTP response, and to extract information from the HTTP response.

`````python
# import beautifulsoup4
import bs4 # should be pre installed with anacondas
`````

### Using BS4 to parse html data

`````python
# we had the content from before stored in the variable my_response
print(my_response.content)
`````

`````python
# bs4.BeautifulSoup # This is a class (a form of function) which handles html data
my_soup = bs4.BeautifulSoup(my_response.content)
`````

`````python
# what is this "my_soup" object?
type(my_soup)
#print(my_soup)
`````

### Methods for the "BeautifulSoup" object

* `.find(tag)` returns the FIRST specified tag
* `.find_all(tag)` returns ALL specified tags


`````python
# this will find the FIRST tag "a" (i.e. the first link)
my_soup.find("a")# this will find ALL tags "a" (i.e. all links)
`````

`````python
# this will find ALL tags "a" (i.e. all links) - returns a list of tags
my_soup.find_all("a")
`````

`````python
# .find(tag) returns a TAG object, .find_all(tag) returns a list of TAG objects
tag_found = my_soup.find("a")
type(tag_found)
# print(tag_found)
`````

### Methods for the `tag` object
* `tag_object.get(attribute)` returns the specified attribute of the tag
* `tag_object.text` returns the text of the object (if it exists)


`````python
print(tag_found)
`````

`````python
tag_found.get('href')
`````

`````python
tag_found.text
`````

## Top trick
You can specify the type of class in `find_all` to filter this a bit.

Looking at ITU's webpage we see that the coffee is stored in the tag: `<div class="col-xs-4">`

`````python
my_soup.find_all('div',{'class':'col-xs-4'}) # # now we are only left with fewer tags
`````

`````python
coffee_tags = my_soup.find_all('div',{'class':'col-xs-4'}) # lets store those in a variable
`````

`````python
coffee_tags[1].text # this looks strange
`````

`````python
coffee_tags[1] # text is stored in two tags
`````

`````python
print(coffee_tags[1].find('h2').text)
print(coffee_tags[1].find('p').text)
`````

`````python
# we can print all the things using a list comprehension
['{} {}'.format(t.find('h2').text,t.find('p').text ) for t in coffee_tags]
`````
