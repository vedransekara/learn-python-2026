# Debugging...
### is the process of finding the root cause, workarounds, and possible fixes for bugs in code

#### What is the issue here?

````python
def multiply(a, b):
    result = a * b

answer = multiply(3, 4)
print(answer)
````

#### What about this one?

````python
my_list = [1, 2, 3, 4, 5]

for i in range(len(my_list)):
    if my_list[i] % 2 == 0:
        my_list[i] += 1

print("Modified list:", my_list)
````

# <font color='red'> Top tip </font>
### Modifying a list while iterating over it can lead to unexpected behavior and errors, so avoid it '_like it's the plague_'

#### A better example of what the previous code was trying to do is:

Old code 👇
```python
for i in range(len(my_list)):
    if my_list[i] % 2 == 0:
        my_list[i] += 1
```

````python
# better approach
my_list = [1, 2, 3, 4, 5]
copy_of_list = my_list.copy()

for item in copy_of_list:
    if item % 2 == 0:
        my_list.remove(item)

print(my_list)  # Output: [1, 3, 5]
````

### Or even better - using a list comprehension 😇

````python
my_list = [1, 2, 3, 4, 5]
my_list = [item for item in my_list if item % 2 != 0]
print(my_list)
````

# Recap of mutable/immutable objects

| Feature |	Mutable Objects | Immutable Objects |
| ---- | ---- | ----- |
| Can be modified after creation | Yes | No |
| Object identity remains the same after modification | Yes | No (a new object is created when changed) |
| Examples | list, dict, set | int, float, bool, str, tuple |
| Suitable for | Data that changes frequently | Fixed or constant data |

### My favourite built-in function, which we havent talked about yet, `zip()`

````python
names = ['Alice','Bob','Charlie','Douglas','Ellis']
grades = ['12','7','02','4','10']

# how do we combine two or more lists (or tuples, strings, dictionaries, etc.) into a single list of tuples? 
# we want = [('Alice', '12'), ('Bob', '7'), ('Charlie', '02'), ('Douglas', '4'), ('Ellis', '10')]
````

### Finding the Highest Value in a dictionary
```python
sales = {
    "January": 4500,
    "February": 5200,
    "March": 4100,
    "April": 6100
}
```

There are many solutions to this

````python
# define the dict
sales = {
    "January": 4500,
    "February": 5200,
    "March": 4100,
    "April": 6100
}

max_v = 0 # given the data above this seems like a good reference point, could also have been max_v = -9999999
max_k = ''
for k,v in sales.items():
    if v > max_v:
        max_v = v
        max_k = k
print('max value: {}, max key: {}'.format(max_v,max_k))
````

__Another solution using `max` & `lamdba` functions__

````python
sales = {
    "January": 4500,
    "February": 5200,
    "March": 4100,
    "April": 6100
}

max(sales.items(),key=lambda x: x[1])
````

### Filtering dictionary items

```python
employees = {
    "John": 45000,

```

__Filter by only keeping employees earning more than 60,000.__

````python
employees = {
    "John": 45000,
    "Anna": 75000,
    "Mark": 62000,
    "Sara": 84000
}

filtered_dict = dict()
for k,v in employees.items():
    if v > 60000:
        filtered_dict[k] = v 

print(filtered_dict)
````

## merging two dictionaries
```python
inventory1 = {
    "Laptop": 10,
    "Mouse": 25
}

inventory2 = {
    "Keyboard": 15,
    "Monitor": 8
}
```

__There are multiple approaches: you can__
1. __keep one dictionary and merge the other into it__
2. __or create a new dict to merge the two together__

````python
inventory1 = {
    "Laptop": 10,
    "Mouse": 25
}

inventory2 = {
    "Keyboard": 15,
    "Monitor": 8
}

merged_inventory = inventory1.copy()
for k,v in inventory2.items():
    if k in merged_inventory:
        merged_inventory[k] += v
    else:
        merged_inventory[k] = v
````

````python
merged_inventory
````

### The super secret solution (which we did not talk about in class - until now)
#### The `.update()` method can be used to merge dictionaries. It modifies one dictionary by adding or updating key-value pairs from another.

````python
merged_inventory = inventory1.copy()
merged_inventory.update(inventory2)
````

````python
merged_inventory
````
